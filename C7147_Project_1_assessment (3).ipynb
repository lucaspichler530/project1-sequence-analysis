## 🧬 Load reference and reads data into variables
Below, we provide some helper code to load the reference from `reference.txt` and the reads from `reads.csv`.

❗ **Do not change the function’s behavior** (you may read it, of course!).

The code creates two variables from the loaded files:
- **reference**: a string containing the 200-nt reference sequence
- **reads**: a list of dictionaries containing the 6,000 reads

````python
##### Constants

REF_PATH = "reference.txt"
READS_PATH = "reads.csv"

##### Functions

def load_reference(path: str) -> str:
    """Load reference nucleotide sequence from file at path to reference string."""
    with open(path) as f:
        lines = [ln.strip() for ln in f if ln.strip()]
    if lines[0].startswith('>'):
        lines = lines[1:]
    ref = "".join(lines).upper()
    if len(ref) != 200:
        raise ValueError(f"Reference must be 200 nt, got {len(ref)}")
    return ref


def load_reads(path: str) -> list:
    """Load sample NGS reads from file at path to reads list."""
    with open(path, newline="") as f:
        reader = csv.DictReader(f)
        reads = list(reader)
    return reads

##### Main code

# Load reference and reads
reference = load_reference(REF_PATH)
reads = load_reads(READS_PATH)

# Report data structures
print("Reference length:", len(reference))
print("Total no. of reads:", len(reads))
print("Keys in each read dictionary:", reads[0].keys())
print("Samples loaded:", sorted({read["sample"] for read in reads}))

# Print sample sequences
print("\nSample read data:")
for row in reads[:5]:
    print(row["read_id"], row["sample"], row['group'], row["sequence"])
print("...")
for row in reads[-5:]:
    print(row["read_id"], row["sample"], row['group'], row["sequence"])
````

## Step 1 — Validate each read and record any errors
## Function: `validate_read(sequence, reference)`

### 👍 You can assume:
- `reference` is a non-empty string containing only uppercase A/C/G/T.
- `sequence` is a string, but nothing about its length or contents.

### 🎯 Aims:
- Check that an individual read is the same length as the reference and contains only valid characters (A/C/G/T)
- "Repair" any reads that contain valid lower-case characters -> upper-case
- Return a dictionary with the following keys:

    - `sequence`: the upper-case version of the sequence input, including when invalid.
    - `valid`: `True` only when the length matches the reference and every character is A/C/G/T.
    - `errors`: a list of strings. Include `"wrong_length"` first if lengths differ, then `"invalid_characters"` if any character is outside A/C/G/T. Include both when both apply. A valid read has `[]`.

### 🔎 Expected outputs

Examples for reference `"ATGC"`:
| Input (read) | Input type | Output |
|---|---|:---|
| `"ATGC"` | Correct | `{"sequence": "ATGC", "valid": True, "errors": []}` |
| `"atgc"` | Lowercase | `{"sequence": "ATGC", "valid": True, "errors": []}` |
| `""` | Empty read | `{"sequence": "", "valid": False, "errors": ["wrong_length"]}` |
| `"ATNC"` | Invalid character | `{"sequence": "ATNC", "valid": False, "errors": ["invalid_characters"]}` |
| `"ATN"` | Multiple errors | `{"sequence": "ATN", "valid": False, "errors": ["wrong_length", "invalid_characters"]}` |


Invalid reads should not crash your program.

````python
validate_read("ATGC", "ATGC")
````

### 💻 Enter your code here

````python
def validate_read(sequence, reference):
    """ Checks whether a DNA sequence has the correct length, and contains only valid DNA characters (A, T, C and G).

    It will first convert lowercase characters to uppercase.

    It will finally also return a dictionary containing the sequence, whether it is valid, and any errors.

    Parameters:
        sequence (str): The DNA sequence to validate.
        reference (str): The reference DNA sequence to be compared with.

    Returns:
        dict: Contains uppercase sequence, a Boolean indicating whether it is valid, and a list of errors. """

  # The method .upper() converts the sequence added to uppercase
    sequence = sequence.upper()

    errors = []  # The use of [] creates an empty list, to store the number of errors in this example

  # The if statement and len() function checks whether the sequence has the correct length
  # The method .append() adds a single element of any data type to the end of a list
    if len(sequence) != len(reference):
      errors.append("wrong_length")

  # The for loop and if statement checks whether the sequence contains any invalid characters
    for base in sequence:
      if base not in "ATCG":
        errors.append("invalid_characters")
        break  # Stops the for loop after finding the first error

  # The return statement stores the values in a dictionary for future use
    return {
        "sequence": sequence,
        "valid": len(errors) == 0,  # Checks whether the sequence is valid by counting the errors
        "errors": errors
    }
````

## Step 2 — Find every mismatch in a valid read
## Function: `find_mutations(sequence, reference)`

### 👍 You can assume:
- Inputs `sequence` and `reference` are valid (i.e. both are strings of the same length and contain only upper-case bases, i.e. A/C/G/T)

### 🎯 Aims:
- Return a list containing one dictionary per mismatch, in ascending position order.
- Each dictionary has exactly these keys:
    - `position`: position of the mutation (integer), numbered from 1.
    - `reference_base`: one-character string from reference at that position.
    - `read_base`: one-character string from the read at that position.

- Return `[]` if the read matches the reference.

### 🔎 Expected outputs

Example: `find_mutations("GTGA", "ATGC")` returns:
```python
[{"position": 1, "reference_base": "A", "read_base": "G"},
 {"position": 4, "reference_base": "C", "read_base": "A"}]
```


### 💻 Enter your code here

````python
def find_mutations(sequence, reference):
    """ Compares a DNA sequence with a reference sequence, and identifies every position where they differ.

    Returns a list of dictionaries containing the position, and the reference and sequence base at that position.

    Parameters:
        sequence (str): The DNA read to compare.
        reference (str): The reference DNA to be compared with.

    Returns:
        list: Dictionaries containing the mutation position, reference nucleotide and observed nucleotide """

    mutations = []  # [] creates an empty list, to store the number of mutations in this example

  # The for loop and range(len()) function creates a loop comparing the exact base-by-base positions in the sequence and reference DNA
  # The method .append({}) adds a dictionary to the end of the 'mutations' list
  # The return statement stores the completed list into 'mutations' for future use
    for i in range(len(reference)):
      if sequence[i] != reference[i]:
        mutations.append({
            "position": i + 1,
            "reference_base": reference[i],
            "read_base": sequence[i]
        })
    return mutations
````

## Step 3 — Analyse all reads
## Function: `analyse_reads(reads, reference)`

### 👍 You can assume:
- `reads` is a list of dictionaries containing keys `read_id`, `sample`, `group` and `sequence`, each with string values.
- `reference` is a non-empty string containing only A/C/G/T.

### 🎯 Aims:
- Use `validate_read` and `find_mutations` to return a **new list of dictionaries containing information about all reads** (even invalid ones, one dictionary per read), preserving the input read order.
- Each read dictionary should contain the following keys (values below are examples):
```python
{
    "read_id": "read_0001",
    "sample": "sample1",
    "group": "treatment",
    "sequence": "GTGA",             # normalised upper-case sequence
    "status": "valid",              # "valid" or "invalid"
    "errors": [],                   # validation error codes
    "mutations": [
        {"position": 1, "reference_base": "A", "read_base": "G"},
        {"position": 4, "reference_base": "C", "read_base": "A"}
    ]
}
```
- Do not change the original read dictionaries.

### 🔎 Expected outputs
- Valid reads have `status="valid"` and `mutations` listing all mutations in the dictionary format shown above.
- Invalid reads have `status="invalid"`, the relevant error codes, and `mutations=[]`.
- Empty input returns `[]`.

### 💻 Enter your code here

````python
def analyse_reads(reads, reference):
    """ Analyses all DNA reads by checking their validity and identifies mutations only in valid reads.

    Returns a new list of dictionaries containing information on the validity and mutations of every read.

    Parameters:
        reads (list): A list of dictionaries containing a read ID, sample name, groups and DNA sequences.
        reference (str): The reference DNA sequence to be compared with.

    Returns:
        list: A new list of dictionaries containing the original read info, validation status, errors, and identified mutations. """

    processed_reads = []  # [] creates an empty list, to store the results of the analysed reads

  # The for loop creates a loop analysing each DNA read individually
  # The validate_read(read[]) function extracts each sequence from the dictionary and checks its validity against the reference using the step 1 function
  # The validation[] operation stores the dictionary of DNA sequence information processed by the validate_read() function into the variable 'sequence'
    for read in reads:
        validation = validate_read(read["sequence"], reference)
        sequence = validation["sequence"]

      # The if statement and find_mutations() function identifies mutations only if the DNA read is valid using the step 2 function
      # The else statement works with the invalid reads which were ignored by the if statement
        if validation["valid"]:
            mutations = find_mutations(sequence, reference)
            validity = "valid"
        else:
            mutations = []  # [] creates an empty list to store the mutations
            validity = "invalid"

      # {} creates a new dictionary as a copy with the relevant information from the first dictionary
        evaluated_reads = {
            "read_id": read["read_id"],
            "sample": read["sample"],
            "group": read["group"],
            "sequence": sequence,
            "status": validity,
            "errors": validation["errors"],
            "mutations": mutations
        }

      # The method .append() adds the new copy dictionary to the 'analysed_reads' results
        processed_reads.append(evaluated_reads)

  # The return statement returns the completed list to 'processed_reads'
    return processed_reads
````

````python
def summarise_samples(analyse_reads):
  """ Summarises DNA mismatch patterns for each sample.

  Calculates the number of valid and invalid reads, the percentage of mistmatches, the total mismatch events, substitution counts and mutation positions.

  Returns a dictionary containing the results for every sample, and a summary table for each group.

  Parameters:
      analysed_reads (list): A list of analysed read dictionaries returned by 'analyse_reads()'

  Returns:
      dict: A dictionary containing mutation summaries for each sample, containing group info, mismatch statistics, and substitution counts. """

# The if statement checks whether or not the input list is empty
  if not analyse_reads:
    return {}  # If the list is empty {} returns an empty dictionary, if else the function continues

  sample_data = {}  # {} creates an empty dictionary

# The for loop creates a loop going through every analysed read
# The read[] operation extracts the 'sample' and 'group' dictionary keys for each read
  for read in analyse_reads:
    sample = read["sample"]
    group = read["group"]

  # The if statement checks whether the sample already exists in the dictionary
    if sample not in sample_data:
      sample_data[sample] = {
          "group": group,
          "valid_reads": 0,
          "invalid_reads": 0,
          "mismatch_reads": 0,
          "mismatch_events": 0,
          "substitution_counts": {},
          "position_counts": {}
      }

    sample_information = sample_data[sample]

  # The if statement checks whether the read is valid
    if read["status"] == "valid":
      sample_information["valid_reads"] += 1  # += 1 increases the 'valid_reads' count by 1 only if the read is valid

      mutations = read["mutations"]

    # The if statement checks whether the valid read contains any mutations
      if mutations:
        sample_information["mismatch_reads"] += 1  # += 1 increases the 'mismatch_reads' count by 1 if the read contains at least 1 mutation

    # The for loop creates a loop going through every mutation in the read
      for mutation in mutations:
        sample_information["mismatch_events"] += 1  # += 1 increases the 'mismatch_events' count by 1 for every individual mutation found in the read

        substitution = (  # A label describing the type of DNA substitution is created
            mutation["reference_base"] + ">" +
            mutation["read_base"]
        )

      # The if statement checks whether the substitution already exists in the dictionary
        if substitution not in sample_information["substitution_counts"]:
          sample_information["substitution_counts"][substitution] = 0  # The substitution type is created with a count of 0
        sample_information["substitution_counts"][substitution] += 1  # += 1 increases the 'substitution_counts' by 1 for each type of substitution found

        position = mutation["position"]

      # The if statement checks whether the position of the substitution already exists in the dictionary
        if position not in sample_information["position_counts"]:
          sample_information["position_counts"][position] = 0  # The substitution position is created with a count of 0
        sample_information["position_counts"][position] += 1  # += 1 increases the 'substitution_position' by 1 for the exact position within a read for every mutation found

  # The else statement increases the 'invalid_reads' count by 1 only if the read is invalid
    else:
      sample_information["invalid_reads"] += 1

  results = {}  # {} creates a new dictionary to store the results

# The for loop creates a loop going through each analysed sample and its information
  for sample, info in sample_data.items():
    valid_reads = info["valid_reads"]

  # The if statement calculates the percentage of valid reads which contain mutations
    if valid_reads > 0:
      mismatch_percentage = (  # Mismatch percentage = (Mismatch reads / Valid reads) * 100
          info["mismatch_reads"] / valid_reads
      ) * 100
    else:  # The else statement uses 'None' if there are no valid reads
      mismatch_percentage = None

  # The sorted([]) operation sorts mutation positions based on frequency
    sorted_positions = sorted(
        info["position_counts"],
        key = lambda position: (
            -info["position_counts"][position], position  # The -info[position_counts] operation sorts mutation frequency by highest to lowest
        )
    )

    top_positions = sorted_positions[:3]  # [:3] takes the first 3 items from the sorted list

    results[sample] = {  # {} creates a final summary dictionary for each sample
        "group": info["group"],
        "valid_reads": valid_reads,
        "invalid_reads": info["invalid_reads"],
        "mismatch_reads": info["mismatch_reads"],
        "mismatch_percentage": mismatch_percentage,
        "mismatch_events": info["mismatch_events"],
        "substitution_counts": dict(info["substitution_counts"]),  # The dict() function ensures the counts are stored as dictionaries
        "position_counts": dict(sorted(info["position_counts"].items())),  # The sorted() function and .items() method sorts the position counts in numerical order
        "top_positions": top_positions
    }

  groups = {}  # {} creates an empty dictionary

# The for loop creates a loop which goes through each sample stored in the 'results' dictionary
  for sample, info in results.items():  # The .items() method accesses both the keys and values in the dictionary
    group = info["group"]

  # The if statement checks whether the group already exists in the dictionary
    if group not in groups:
      groups[group] = []
    groups[group].append(sample)  # The .append() method adds the sample name to the end of its appropriate group list

# The for loop creates a loop going through each group separately
  for group, samples in groups.items():  # The .items() method allows access to both the key and values of the dictionary
    print(f"{'\n' + '=' * 70}")  # '\n' creates a new line
    print(f"Group: {group}")
    print(f"{'=' * 70}")  # '"=" * 70' prints 70 equals signs
    print(f"\nSummary of Mismatch Patterns:")
    print(
        f"{'Sample':<12} {'Valid':<8} {'Invalid':<9} "  # '<12', '<8' and '<9' control the spacing of the column headings
        f"{'Mismatch %':<12} {'Events':<9} {'Top positions'}"
    )

  # The for loop creates a loop goign through the sample's calculated results
    for sample in samples:
      information = results[sample]

    # The if statement checks whether a percentage was calulated for a read (valid or invalid) and converts 'None' to 'N/A'
      if information["mismatch_percentage"] is None:
        percentage = "N/A"
      else:  # The else statement and ':.2f' rounds the percentage to two decimal places
        percentage = f"{information['mismatch_percentage']:.2f}%"

      print(
          f"{sample:<12} "
          f"{information['valid_reads']:<8} "
          f"{information['invalid_reads']:<9} "
          f"{percentage:<12} "
          f"{information['mismatch_events']:<9} "
          f"{information['top_positions']}"
      )

  # The for loops find every different substitution which occurred in the group
    substitutions = sorted(set(  # The set() function removes duplicates - The sorted() function sorts the substitutions in alphabetical order
        substitution
        for sample in samples
        for substitution in results[sample]["substitution_counts"]
    ))

    print(f"\nSubstitution Mutations:")
    print(f"Each cell shows count (percentage per 100 valid reads).")

  # The if statement prints a message if there are no mutations in the group
  # The end = "" operations keep every print statement on the same line
    if not substitutions:
      print(f"No substitutions observed.")
    else:  # The else statement prints the substitution table if there are substitutions
      print(f"{'Substitution':<16}", end = "")

    # The for loop creates a loop printing the sample names as headings
      for sample in samples:
        print(f"{sample:<20}", end = "")

      print()

    # The for loop creates a loop going through every type of substitution
      for substitution in substitutions:
        print(f" {substitution:<15}", end = "")

      # The for loop creates a loop checking whether there was a substitution observed in the sample
        for sample in samples:
          information = results[sample]
          count = information["substitution_counts"].get(  # The .get() method collects the count for that substitution type in the sample
              substitution, 0
          )

        # The if statement calculates the percentage of valid reads which contain that specific type of substitution
          if information["valid_reads"] > 0:
            percentage = (  # Percentage of reads = (Count for type of substitution / Valid reads) * 100
                count / information["valid_reads"]
            ) * 100

            value = f"{count} ({percentage:.2f}%)"  # ':.2f' rounds the percentage to two decimal places
          else:  # The else statement stores 'N/A' in 'value' if there was no percentage calculated (invalid read)
            value = f"{count} (N/A)"

          print(f"{value:<20}", end = "")

        print()

# The return function returns the completed dictionary containing all sample summaries
  return results
````

It returns `None` values when the comparison cannot be calculated.

````python
#@title 📊 calculate_statistics function

from scipy import stats
import math
import warnings


def calculate_statistics(control_percentages, treated_percentages):
    """Compare independent biological samples using a two-sided Welch's t-test."""
    unavailable = {"statistic": None, "p_value": None}

    # Require at least two samples per group and finite numerical values.
    if len(control_percentages) < 2 or len(treated_percentages) < 2:
        return unavailable

    values = list(control_percentages) + list(treated_percentages)
    if any(value is None or not math.isfinite(value) for value in values):
        return unavailable

    # No within-group variation: the t-test cannot estimate uncertainty.
    if (
        len(set(control_percentages)) == 1
        and len(set(treated_percentages)) == 1
    ):
        return unavailable

    with warnings.catch_warnings():
        warnings.simplefilter("ignore", RuntimeWarning)
        result = stats.ttest_ind(
            treated_percentages,
            control_percentages,
            equal_var=False,
            alternative="two-sided"
        )

    if not (
        math.isfinite(result.statistic)
        and math.isfinite(result.pvalue)
    ):
        return unavailable

    return {
        "statistic": float(result.statistic),
        "p_value": float(result.pvalue)
    }
````

### 💻 Enter your code here

````python
def compare_positions(sample_summary, control_group = "control", treated_group = "treatment"):
  """ Compares the three most commonly mutated DNA positions between the control and treatment groups.

  Calculates mutation percentages for individual samples and compares the groups using Welch's t-test.

  Finally, returns a list of dictionaries containing the results.

  Parameters:
      sample_summary (dict): A dictionary containing sample summaries returned by 'summarise_samples()'.
      control_group (str): The name of the 'control' group.
      Treated_group (str): The name of the 'treatment' group.

  Returns:
      list: Dictionaries containing the selected mutation positions, sample numbers, group means, differences, t-test statistics, and p-values. """

  combined_counts = {}  # {} creates an empty dictionary named 'combined_counts'

# The for loop creates a loop going through every sample in the 'sample_summary' dictionary
  for sample, information in sample_summary.items():  # The .items() method allows access to the keys and values of the dictionary

  # The if statement checks whether or not the sample belongs in the 'control' or 'treatment' group
  # The continue statement allows for the if statement to skip this part of the code
    if information["group"] not in [control_group, treated_group]:
      continue

  # The for loop creates a loop going through the mutation positions recorded in each sample
    for position, count in information["position_counts"].items():  # The .items() method allows access to the keys and values of the dictionary

    # The if statement checks whether the position already exists within the dictionary
      if position not in combined_counts:
        combined_counts[position] = 0  # The position is created with a starting count of 0

      combined_counts[position] += count  # += adds the current sample's mutation count to the existing total

  sorted_positions = sorted(
      combined_counts,
      key = lambda position: (
          -combined_counts[position], position  # The -combined_counts[position] operation sorts positions from highest to lowest
      )
  )

  top_positions = sorted_positions[:3]  # [:3] means take the first three items from the sorted list

  results = []  # [] creates an empty list called 'results' to store the results

# The for loop creates a loop analysing each individual DNA position
# [] creates two empty lists for each position for the 'control' and 'treatment' samples
  for position in top_positions:
    control_percentages = []
    treated_percentages = []

  # The for loop creates a loop going through every sample another time
    for sample, info in sample_summary.items():  # The .items() method allows access to the keys and values of the dictionary
      group = info["group"]  # It retrieves the sample's group
      valid_reads = info["valid_reads"]  # It retrieves the number of valid reads in the sample

    # The if statement checks whether the sample belongs in either the 'control' or 'treatment' group
      if group not in [control_group, treated_group]:
        continue

    # The if statement checks whether the sample has any valid DNA reads
      if valid_reads == 0:
        print(
            f"Excluded {sample} due to no valid reads "
            f"for position {position}."  # It prints out why the sample was excluded
        )
        continue

    # The .get(position, 0) method retrieves the number of mutations at that specific position of the DNA sample
      mutation_count = info["position_counts"].get(position, 0)

      percentage = (mutation_count / valid_reads) * 100  # Percentage = (Mutation count / Valid reads) * 100

    # The if statement and .append() method place the appropriate calculated percentage 'control' group
      if group == control_group:
        control_percentages.append(percentage)

    # The elif statement and .append() method place the appropriate calculated percentage in the 'treatment' group
      elif group == treated_group:
        treated_percentages.append(percentage)

  # The if statement calculates the mean mutation percentage across all the 'control' samples
    if control_percentages:
      control_mean = sum(control_percentages) / len(control_percentages)
    else:  # The else statement sets the result to 'None' if there are no valid 'control' samples
      control_mean = None

  # The if statement calculates the mean mutation percentage across all the 'treatment' samples
    if treated_percentages:
      treated_mean = sum(treated_percentages) / len(treated_percentages)
    else:  # The else statement sets the result to 'None' if there are no valid 'treatment' samples
      treated_mean = None

  # The if statement calculates the difference between the 'control' and 'treatment' group means
    if control_mean is not None and treated_mean is not None:
      difference = treated_mean - control_mean  # Difference = Treatment mean - Control mean
    else:  # The else statement sets the difference to 'None' if there are no valid reads in either group
      difference = None

  # The statistical_result function performs a Welch's t-test on the two 'control' and 'treatment' percentage lists
    statistical_result = calculate_statistics(
        control_percentages,
        treated_percentages
    )

  # {} creates a dictionary containing all the results for one specific DNA position
    position_result = {
        "position": position,
        "control_n": len(control_percentages),
        "treated_n": len(treated_percentages),
        "control_mean": control_mean,
        "treated_mean": treated_mean,
        "difference": difference,
        "statistic": statistical_result["statistic"],
        "p_value": statistical_result["p_value"]
    }

    results.append(position_result)  # The .append() method adds the dictionary for every specific position to the results list

  # The if statement and ':.2f' rounds the 'control' group mean to two decimal places if it exists
    control_display = (
        f"{control_mean:.2f}%"

        if control_mean is not None else "N/A"
    )

  # The if statement and ':.2f' rounds the 'treatment' group mean to two decimal places if it exists
    treated_display = (
        f"{treated_mean:.2f}%"

        if treated_mean is not None else "N/A"
    )

  # The if statement checks whether the 'p_value' is 'None' and therefore could not be calculated
    if statistical_result["p_value"] is None:
      print(
          f"Position {position}: "
          f"{control_group} mean = {control_display} "
          f"(n = {len(control_percentages)}), "
          f"{treated_group} mean = {treated_display} "
          f"(n = {len(treated_percentages)}); "
          "Welch's t-test could not be calculated."
      )
    else:  # The else statement prints the results of the 'p_value'
      print(
          f"Position {position}: "
          f"{control_group} mean = {control_display} "
          f"(n = {len(control_percentages)}), "
          f"{treated_group} mean = {treated_display} "
          f"(n = {len(treated_percentages)}); "
          f"difference = {difference:.2f} percentage points; "
          f"Welch's t = {statistical_result['statistic']:.3f}, "
          f"p = {statistical_result['p_value']:.4f}."
      )

# The return function returns the completed list of dictionaries
  return results
````
